# Evaluation Metrics

To comprehensively evaluate the performance of the classification models, multiple evaluation metrics will be used instead of relying solely on accuracy. Since the customer churn dataset exhibits moderate class imbalance, accuracy alone may not provide a complete picture of model performance.

The following metrics will be considered:

### Accuracy
Accuracy measures the proportion of correctly classified observations out of all observations.

**Use Case:** Provides an overall measure of model performance but can be misleading when classes are imbalanced.

---

### Precision
Precision measures the proportion of customers predicted to churn that actually churned.

**Use Case:** Important when false positives are costly, such as unnecessarily targeting customers with retention offers.

---

### Recall (Sensitivity)
Recall measures the proportion of actual churning customers that the model correctly identifies.

**Use Case:** Particularly important in customer churn prediction because failing to identify a customer who is about to leave (false negative) may result in losing that customer permanently.

---

### F1-Score
The F1-score is the harmonic mean of Precision and Recall.

**Use Case:** Provides a balanced evaluation when both false positives and false negatives are important.

---

### ROC-AUC Score
The Receiver Operating Characteristic - Area Under the Curve (ROC-AUC) evaluates the model's ability to distinguish between churning and non-churning customers across all possible classification thresholds.

**Use Case:** A higher ROC-AUC indicates better overall discrimination capability regardless of the chosen probability threshold.

---

### Confusion Matrix
A confusion matrix summarizes the model's predictions by reporting:

- True Positives (TP)
- True Negatives (TN)
- False Positives (FP)
- False Negatives (FN)

**Use Case:** Helps understand the types of prediction errors made by the model.

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
```

```python
#from now on will use this function to reduce redundant metrics code block
def evaluate_model(model, X_train, y_train, X_test, y_test):
    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    train_prob = model.predict_proba(X_train)[:, 1]
    test_prob = model.predict_proba(X_test)[:, 1]

    metrics = {
        "Accuracy": (
            accuracy_score(y_train, train_pred),
            accuracy_score(y_test, test_pred)
        ),
        "Precision": (
            precision_score(y_train, train_pred),
            precision_score(y_test, test_pred)
        ),
        "Recall": (
            recall_score(y_train, train_pred),
            recall_score(y_test, test_pred)
        ),
        "F1-Score": (
            f1_score(y_train, train_pred),
            f1_score(y_test, test_pred)
        ),
        "ROC-AUC": (
            roc_auc_score(y_train, train_prob),
            roc_auc_score(y_test, test_prob)
        )
    }

    print(f"{'Metric':<12} | {'Training':<10} | {'Testing':<10}")
    print("-" * 40)

    for metric, (train_score, test_score) in metrics.items():
        print(f"{metric:<12} | {train_score:<10.4f} | {test_score:<10.4f}")

    gap = metrics["ROC-AUC"][0] - metrics["ROC-AUC"][1]

    print(f"\nDiagnostic: Train/Test ROC-AUC Gap = {gap:.4f}")

    if gap > 0.05:
        print("Possible overfitting detected.")
    else:
        print("Model is generalizing well.")
```

# Performance After Applying SMOTE

After applying SMOTE, the overall **accuracy** of the Logistic Regression model decreases slightly. This is an expected outcome and should not be interpreted as a deterioration in model performance.

The original dataset was moderately imbalanced, with non-churning customers forming the majority class. Consequently, the baseline model tended to favor predicting the majority class, resulting in high overall accuracy but comparatively lower Recall for the minority (churn) class.

By balancing the training data using SMOTE, the model is exposed to a more representative distribution of both classes. As a result:

- **Recall increases**, indicating that the model correctly identifies a greater proportion of customers who are likely to churn.
- **F1-score improves**, reflecting a better balance between Precision and Recall.
- **ROC-AUC remains competitive or improves**, demonstrating stronger discrimination between churners and non-churners.
- **Accuracy may decrease slightly** because the model now predicts more customers as potential churners, leading to a modest increase in false positives.

From a business perspective, this trade-off is desirable. In customer churn prediction, failing to identify a customer who is likely to leave (false negative) is generally more costly than incorrectly flagging a loyal customer as being at risk (false positive). Therefore, improving the model's ability to detect churners is more valuable than maximizing overall accuracy.

Overall, the SMOTE-based model provides a more balanced and practically useful classifier for customer churn prediction.

# Elastic Net Logistic Regression

Elastic Net Logistic Regression extends the standard Logistic Regression model by combining both **L1 (Lasso)** and **L2 (Ridge)** regularization. This hybrid regularization helps prevent overfitting while also performing feature selection.

```python
elastic_lr = LogisticRegression(
    penalty="elasticnet",
    l1_ratio=0.6,
    solver = "saga",
    C=0.7,
    max_iter=5000,
    random_state=42
)

elastic_lr.fit(X_train_smote, y_train_smote)
```

```python
train_pred_en = elastic_lr.predict(X_train_smote)
test_pred_en = elastic_lr.predict(X_test)

train_prob_en = elastic_lr.predict_proba(X_train_smote)[:, 1]
test_prob_en = elastic_lr.predict_proba(X_test)[:, 1]

metrics_en = {
    "Accuracy": (
        accuracy_score(y_train_smote, train_pred_en),
        accuracy_score(y_test, test_pred_en)
    ),
    "Precision": (
        precision_score(y_train_smote, train_pred_en),
        precision_score(y_test, test_pred_en)
    ),
    "Recall": (
        recall_score(y_train_smote, train_pred_en),
        recall_score(y_test, test_pred_en)
    ),
    "F1-Score": (
        f1_score(y_train_smote, train_pred_en),
        f1_score(y_test, test_pred_en)
    ),
    "ROC-AUC": (
        roc_auc_score(y_train_smote, train_prob_en),
        roc_auc_score(y_test, test_prob_en)
    )
}

print(f"{'Metric':<12} | {'Training':<10} | {'Testing':<10}")
print("-" * 40)

for metric, (train_score, test_score) in metrics_en.items():
    print(f"{metric:<12} | {train_score:<10.4f} | {test_score:<10.4f}")

gap = metrics_en["ROC-AUC"][0] - metrics_en["ROC-AUC"][1]

print(f"\nDiagnostic: Train/Test ROC-AUC Gap = {gap:.4f}")

if gap > 0.05:
    print("Possible overfitting detected.")
else:
    print("Model is generalizing well.")
```

### Observation

Applying Elastic Net regularization resulted in only marginal changes in the model's performance. The evaluation metrics remained largely similar to those of the standard Logistic Regression model, indicating that regularization did not provide a significant improvement. This suggests that the baseline model was already generalizing well, and the dataset does not strongly benefit from additional regularization.

# Support Vector Machine (SVM)

Support Vector Machine (SVM) is a supervised learning algorithm that classifies data by finding the optimal decision boundary (hyperplane) that maximizes the margin between different classes. By maximizing this margin, SVM often achieves better generalization on unseen data.

In this project, an SVM classifier with the Radial Basis Function (RBF) kernel is used. The RBF kernel enables the model to learn non-linear decision boundaries, making it suitable for datasets where the relationship between features and the target is not strictly linear.

```python
from sklearn.svm import SVC
```

```python
svm = SVC(
    kernel="rbf",
    C=0.1,
    gamma="scale",
    probability=True,
    random_state=42
)

svm.fit(X_train_smote, y_train_smote)
```

```python
evaluate_model(svm, X_train_smote, y_train_smote, X_test, y_test)
```

# K-Nearest Neighbors (KNN)

K-Nearest Neighbors (KNN) is a non-parametric, instance-based learning algorithm that classifies a sample based on the majority class among its *k* nearest neighbors in the feature space. Unlike Logistic Regression or SVM, KNN does not explicitly learn a model during training but instead makes predictions by comparing distances to existing training samples.
Since KNN relies on distance calculations, feature scaling is essential to ensure that all features contribute equally to the distance metric. 

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(
    n_neighbors=50
)

knn.fit(X_train_smote, y_train_smote)
evaluate_model(knn, X_train_smote, y_train_smote, X_test, y_test)
```

### Observation

KNN achieves high Recall but low Precision and Accuracy, resulting in many false positives. Overall, it does not outperform Logistic Regression.

# Decision Tree

Decision Tree is a non-linear supervised learning algorithm that classifies data by recursively splitting it based on the most informative features. It is easy to interpret and can capture complex relationships within the data. However, Decision Trees are prone to overfitting, making careful evaluation essential before selecting them as the final model.

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(
    criterion="gini",
    random_state=42,
    max_depth= 3,
    min_samples_split=15,
)

dt.fit(X_train_smote, y_train_smote)
evaluate_model(dt, X_train_smote, y_train_smote, X_test, y_test)
```

# Random Forest Classifier

Random Forest is an ensemble learning algorithm based on the **bagging** technique. It builds multiple Decision Trees on different bootstrap samples of the training data and combines their predictions through majority voting. By averaging the predictions of several trees, Random Forest reduces overfitting and generally provides better generalization than a single Decision Tree.

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(
     n_estimators=200,
    max_depth=5,
    min_samples_split=20,
    min_samples_leaf=10,
    random_state=42
)

rf.fit(X_train_smote, y_train_smote)
evaluate_model(rf, X_train_smote, y_train_smote, X_test, y_test)
```

# Gradient Boosting Classifier

Gradient Boosting is an ensemble learning algorithm based on the **boosting** technique. It builds Decision Trees sequentially, with each new tree learning from the errors made by the previous ones. By iteratively improving weak learners, Gradient Boosting often achieves higher predictive performance than a single Decision Tree or bagging-based methods.

```python
from sklearn.ensemble import GradientBoostingClassifier
gb = GradientBoostingClassifier(
    n_estimators=200,
    learning_rate=0.001,
    max_depth=2,
    random_state=42
)

gb.fit(X_train_smote, y_train_smote)
evaluate_model(gb, X_train_smote, y_train_smote, X_test, y_test)
```

# XGBoost Classifier


XGBoost (Extreme Gradient Boosting) is an advanced ensemble learning algorithm based on the boosting technique. It builds Decision Trees sequentially, where each tree corrects the errors of the previous ones. Compared to traditional Gradient Boosting, XGBoost incorporates regularization, efficient tree construction, and built-in optimizations, often resulting in superior predictive performance and better generalization on structured datasets.

```python
from xgboost import XGBClassifier
xgb = XGBClassifier(
    n_estimators=100,
    learning_rate=0.01,
    max_depth=3,
    random_state=42,
    colsample_bytree=0.9,
    reg_alpha=0.4,
    reg_lambda=1,
    eval_metric="aucpr"
)

xgb.fit(X_train_smote, y_train_smote)
evaluate_model(xgb, X_train_smote, y_train_smote, X_test, y_test)
```
